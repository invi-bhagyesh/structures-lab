```python
class BankAccount:
    def __init__(self, account_number, owner_name, balance, password):
        self.__account_number = account_number
        self.__owner_name = owner_name
        self.__balance = balance
        self.__password = password

    # Getters and setters for private attributes
    def get_account_number(self):
        return self.__account_number

    def get_owner_name(self):
        return self.__owner_name

    def set_owner_name(self, name):
        self.__owner_name = name

    def get_balance(self):
        return self.__balance

    # Private method to check password
    def __check_password(self, password):
        return password == self.__password

    # Deposit method
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.__balance += amount
        print(f"Deposited {amount}. New balance: {self.__balance}")

    # Withdraw method (password protected)
    def withdraw(self, amount, password):
        if not self.__check_password(password):
            raise PermissionError("Incorrect password.")
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.__balance:
            raise ValueError("Insufficient balance.")
        self.__balance -= amount
        print(f"Withdrew {amount}. New balance: {self.__balance}")

    # Transfer method
    def transfer(self, to_account, amount, password):
        if not self.__check_password(password):
            raise PermissionError("Incorrect password.")
        if amount <= 0:
            raise ValueError("Transfer amount must be positive.")
        if amount > self.__balance:
            raise ValueError("Insufficient balance.")
        self.__balance -= amount
        to_account.__balance += amount
        print(f"Transferred {amount} to account {to_account.get_account_number()}. New balance: {self.__balance}")

    # Display account details (except password)
    def display(self):
        print(f"Account Number: {self.__account_number}")
        print(f"Owner Name: {self.__owner_name}")
        print(f"Balance: {self.__balance}")


class SavingsAccount(BankAccount):
    MIN_BALANCE = 1000
    PENALTY = 100

    def withdraw(self, amount, password):
        super().withdraw(amount, password)
        # Check for minimum balance penalty
        if self.get_balance() < self.MIN_BALANCE:
            self._apply_penalty()

    def _apply_penalty(self):
        balance = self.get_balance()
        if balance < self.PENALTY:
            print("Balance too low to apply penalty.")
            return
        # Reduce balance by penalty
        self._BankAccount__balance -= self.PENALTY
        print(f"Penalty of Rs.{self.PENALTY} applied due to low balance. New balance: {self.get_balance()}")


class CurrentAccount(BankAccount):
    OVERDRAFT_LIMIT = 5000

    def withdraw(self, amount, password):
        if not self._BankAccount__check_password(password):
            raise PermissionError("Incorrect password.")
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.get_balance() + self.OVERDRAFT_LIMIT:
            raise ValueError("Overdraft limit exceeded.")
        self._BankAccount__balance -= amount
        print(f"Withdrew {amount}. New balance: {self.get_balance()}")


# Menu-driven program
def banking_menu():
    accounts = {}

    def create_account():
        acc_type = input("Enter account type (Savings/Current): ").strip().lower()
        acc_num = input("Account Number: ").strip()
        owner = input("Owner Name: ").strip()
        try:
            balance = float(input("Initial Deposit: "))
            password = input("Set Password: ").strip()
            if acc_type == 'savings':
                acc = SavingsAccount(acc_num, owner, balance, password)
            elif acc_type == 'current':
                acc = CurrentAccount(acc_num, owner, balance, password)
            else:
                print("Invalid account type!")
                return
            accounts[acc_num] = acc
            print("Account created successfully.")
        except ValueError:
            print("Invalid input for balance.")

    def deposit_money():
        acc_num = input("Account Number: ").strip()
        if acc_num not in accounts:
            print("Account not found.")
            return
        try:
            amount = float(input("Amount to deposit: "))
            accounts[acc_num].deposit(amount)
        except ValueError as e:
            print(e)

    def withdraw_money():
        acc_num = input("Account Number: ").strip()
        if acc_num not in accounts:
            print("Account not found.")
            return
        password = input("Enter Password: ").strip()
        try:
            amount = float(input("Amount to withdraw: "))
            accounts[acc_num].withdraw(amount, password)
        except Exception as e:
            print(e)

    def transfer_money():
        from_acc = input("Your Account Number: ").strip()
        to_acc = input("Recipient Account Number: ").strip()
        if from_acc not in accounts or to_acc not in accounts:
            print("One or both accounts not found.")
            return
        password = input("Enter your password: ").strip()
        try:
            amount = float(input("Amount to transfer: "))
            accounts[from_acc].transfer(accounts[to_acc], amount, password)
        except Exception as e:
            print(e)

    def display_account():
        acc_num = input("Account Number: ").strip()
        if acc_num not in accounts:
            print("Account not found.")
            return
        accounts[acc_num].display()

    menu_options = {
        '1': create_account,
        '2': deposit_money,
        '3': withdraw_money,
        '4': transfer_money,
        '5': display_account,
        '6': sys.exit
    }

    while True:
        print("\n--- Banking System Menu ---")
        print("1. Create Account")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transfer Money")
        print("5. Display Account Details")
        print("6. Exit")
        choice = input("Choose an option: ").strip()
        action = menu_options.get(choice)
        if action:
            action()
        else:
            print("Invalid option, try again.")


```

```python
from abc import ABC, abstractmethod
import math 
import sys

# Abstract base Shape class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

# Abstract subclass SolidShape
class SolidShape(Shape):
    @abstractmethod
    def volume(self):
        pass

# 2D Shapes

class Rectangle(Shape):
    def __init__(self, length, width):
        if length <= 0 or width <= 0:
            raise ValueError("Length and Width must be positive numbers.")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

class Circle(Shape):
    def __init__(self, radius):
        if radius <= 0:
            raise ValueError("Radius must be positive.")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

# 3D Shapes

class Cylinder(SolidShape):
    def __init__(self, radius, height):
        if radius <= 0 or height <= 0:
            raise ValueError("Radius and Height must be positive.")
        self.radius = radius
        self.height = height

    def area(self):  # Surface Area
        return 2 * math.pi * self.radius * (self.radius + self.height)

    def perimeter(self):
        # Perimeter concept does not apply well to 3D shapes, 
        # but for sake of polymorphism, return circumference of base circle
        return 2 * math.pi * self.radius

    def volume(self):
        return math.pi * self.radius ** 2 * self.height

class Cube(SolidShape):
    def __init__(self, side):
        if side <= 0:
            raise ValueError("Side length must be positive.")
        self.side = side

    def area(self):  # Surface Area
        return 6 * self.side ** 2

    def perimeter(self):
        # Total edge length = 12 edges * side length
        return 12 * self.side

    def volume(self):
        return self.side ** 3

# Polymorphic display function
def display_shape_info(shape: Shape):
    print(f"Shape: {type(shape).__name__}")
    print(f"Area: {shape.area():.2f}")
    print(f"Perimeter: {shape.perimeter():.2f}")
    if isinstance(shape, SolidShape):
        print(f"Volume: {shape.volume():.2f}")
    print("-" * 30)


# Menu-driven program
def shape_menu():
    shapes = []

    def create_shape():
        print("Choose shape type:")
        print("1. Rectangle")
        print("2. Circle")
        print("3. Cylinder")
        print("4. Cube")
        choice = input("Enter choice: ").strip()
        try:
            if choice == '1':
                l = float(input("Length: "))
                w = float(input("Width: "))
                shape = Rectangle(l, w)
            elif choice == '2':
                r = float(input("Radius: "))
                shape = Circle(r)
            elif choice == '3':
                r = float(input("Radius: "))
                h = float(input("Height: "))
                shape = Cylinder(r, h)
            elif choice == '4':
                s = float(input("Side length: "))
                shape = Cube(s)
            else:
                print("Invalid choice.")
                return
            shapes.append(shape)
            print(f"{type(shape).__name__} created successfully.")
        except ValueError as e:
            print(f"Error: {e}")

    def display_shapes():
        if not shapes:
            print("No shapes created yet.")
            return
        for shape in shapes:
            display_shape_info(shape)

    def compare_areas():
        if len(shapes) < 2:
            print("At least two shapes needed to compare.")
            return
        print("Comparing areas of all shapes:")
        areas = [(shape, shape.area()) for shape in shapes]
        areas.sort(key=lambda x: x[1], reverse=True)
        for shape, area in areas:
            print(f"{type(shape).__name__}: Area = {area:.2f}")


            
    menu_options = {
        '1': create_shape,
        '2': display_shapes,
        '3': compare_areas,
        '4': sys.exit
    }


        

    while True:
        print("\n--- Shape System Menu ---")
        print("1. Create Shape")
        print("2. Display Shape Details")
        print("3. Compare Areas")
        print("4. Exit")
        choice = input("Choose an option: ").strip()
        action = menu_options.get(choice)
        if action:
            action()
        else:
            print("Invalid option, try again.")

```
