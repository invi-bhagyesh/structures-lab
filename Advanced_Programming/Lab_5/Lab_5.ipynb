## Q.1 Design a Library Management System using classes and objects to manage books across different branches.  
**(a)** Create a class Library with the following:   
- A class attribute library_name (same for all branches).   
- A class attribute total_books to track the total number of books across all branches. 
- Instance attributes: branch_name and books (list of books in that branch).       

**(b)** Implement methods:     
- add_book(book_title) → Adds a book to the branch and updates the total book count.   
- display_books() → Displays all books in that branch.   
- display_total_books() → A class method that shows the total number of books across all branches. 

**(c)** Create two or more library branches and demonstrate adding books, displaying books per     
branch, and showing the total book count. 

```python
class Library:
    library_name = "INVI"
    total_books = 0
    all_books= []
    def __init__(self, branch_name):
        self.branch_name = branch_name
        self.books = []
    
    def add_book(self,book_title):
        self.books.append(book_title)
        Library.all_books.append(book_title)
        Library.total_books +=1
    
    def display_books(self):
            return self.books    
            
    @classmethod
    def display_total_books(cls):
        return cls.all_books
```

```python
b1 = Library("Downtown")
b2 = Library("Uptown")

b1.add_book("Python 101")
b1.add_book("Data Science")
b2.add_book("Machine Learning")

print(b1.display_books())
print(b2.display_books())

print(Library.display_total_books())

print(f"Total books: {Library.total_books}")
```

## Q.2 Develop a Student Result System using inheritance to manage students, their subjects, and final results. 
**a.** Create a base class Person with attributes name and roll_no.   
**b.** Create a derived class Student (inherits from Person) with additional attributes
subject_marks (dictionary).   
**c.** Create another derived class Result (inherits from Student) that:   
i. Calculates total marks and percentage.   
ii. Displays pass/fail status (pass if percentage ≥ 40).   
**d.** Do the following: 
i. Create at least three students.   
ii. Enter marks for 5 subjects per student.   
iii. Display their total marks, percentage, and result status.

```python
class Person:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name, roll_no)
        self.subject_marks = {}  # {subject: marks}
    
    def add_marks(self, subject, marks):
        self.subject_marks[subject] = marks
class Result(Student):
    def __init__(self, name, roll_no):
        super().__init__(name, roll_no)
    
    def calculate_result(self):
        total_marks = sum(self.subject_marks.values())
        percentage = total_marks / len(self.subject_marks)
        status = "Pass" if percentage >= 40 else "Fail"
        return total_marks, percentage, status
    
    def display_result(self):
        total, percentage, status = self.calculate_result()
        print(f"Name: {self.name}, Roll No: {self.roll_no}")
        print(f"Marks: {self.subject_marks}")
        print(f"Total: {total}, Percentage: {percentage:.2f}%, Result: {status}")
        print("-" * 40)
```

```python
# Create 3 students
s1 = Result("Alice", 101)
s2 = Result("Bob", 102)
s3 = Result("Charlie", 103)

# Add marks (5 subjects each)
s1.add_marks("Math", 80)
s1.add_marks("Science", 70)
s1.add_marks("English", 65)
s1.add_marks("History", 75)
s1.add_marks("Computer", 85)

s2.add_marks("Math", 35)
s2.add_marks("Science", 40)
s2.add_marks("English", 30)
s2.add_marks("History", 25)
s2.add_marks("Computer", 45)

s3.add_marks("Math", 55)
s3.add_marks("Science", 60)
s3.add_marks("English", 50)
s3.add_marks("History", 65)
s3.add_marks("Computer", 70)

# Display results
s1.display_result()
s2.display_result()
s3.display_result()
```

## Q.3 ) Create an Employee Payroll System to manage employee salaries using multiple inheritance.  
**a.** Create a class Employee with attributes emp_id and emp_name.   
**b.** Create a class Salary with attributes basic, hra, and da, and a method
calculate_salary() to compute gross salary.   
**c.** Create a class Payroll that inherits from both Employee and Salary and includes:   
i. A method display_details() to show the employee details and total salary.   
**d.** Create objects for at least three employees, input their salary details, and display 
their payroll.   
**e**. Demonstrates multiple inheritance where one class inherits from two parent classes.   


```python
class Employee:
    def __init__(self, emp_id, emp_name):
        self.emp_id = emp_id
        self.emp_name = emp_name
class Salary:
    def __init__(self, basic, hra, da):
        self.basic = basic
        self.hra = hra
        self.da = da
    
    def calculate_salary(self):
        return self.basic + self.hra + self.da
class Payroll(Employee, Salary):
    def __init__(self, emp_id, emp_name, basic, hra, da):
        # Call constructors of both parent classes
        Employee.__init__(self, emp_id, emp_name)
        Salary.__init__(self, basic, hra, da)

    def display_details(self):
        gross_salary = self.calculate_salary()
        print(f"Employee ID: {self.emp_id}")
        print(f"Employee Name: {self.emp_name}")
        print(f"Basic: {self.basic}, HRA: {self.hra}, DA: {self.da}")
        print(f"Gross Salary: {gross_salary}")
        print("-" * 40)
```

```python
# Create objects
emp1 = Payroll(101, "Alice", 30000, 8000, 5000)
emp2 = Payroll(102, "Bob", 25000, 6000, 4000)
emp3 = Payroll(103, "Charlie", 40000, 10000, 7000)

# Display payrolls
emp1.display_details()
emp2.display_details()
emp3.display_details()
```
