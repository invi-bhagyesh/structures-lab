## Library Book Management System

```python

class NotEnoughCopiesError(Exception):
    pass

books = {
    "Python Basics": 4,
    "Data Science 101": 2,
    "Machine Learning": 5
}

print("\n=== Library Book Management System ===")
print("Type 'exit' anytime to quit.\n")

while True:
    try:
        book_name = input("Enter book title to borrow: ").strip()

        if book_name.lower() == "exit":
            print("\nExiting system...")
            break

        try:
            copies = int(input("Enter number of copies to borrow: "))
        except ValueError:
            print(" Error: Number of copies must be an integer.\n")
            continue

        try:
            if book_name not in books:
                raise KeyError

            if copies > books[book_name]:
                raise NotEnoughCopiesError

            # Process borrowing
            books[book_name] -= copies
            print(f"Borrowed {copies} copies of '{book_name}'. Remaining: {books[book_name]}\n")

        except KeyError:
            print("Error: Book not found in library.\n")
        except NotEnoughCopiesError:
            print(f" Error: Only {books[book_name]} copies of '{book_name}' are available.\n")

    except Exception as e:
        print(f"Unexpected error: {e}\n")
```

## Online Shopping Cart Simulation

```python


inventory = {
    "Laptop": 50000,
    "Headphones": 1500,
    "Mouse": 700,
    "Keyboard": 1200
}

cart = []  # List of tuples (item_name, quantity, total_price)

print("\n=== Online Shopping Cart ===")
print("Type 'exit' anytime to checkout.\n")

while True:
    item = input("Enter item name to purchase: ").strip()

    if item.lower() == "exit":
        break

    try:
        if item not in inventory:
            raise KeyError

        try:
            quantity = int(input("Enter quantity: "))
        except ValueError:
            print(" Error: Quantity must be a number.\n")
            continue

        price = inventory[item] * quantity
        cart.append((item, quantity, price))
        print(f"Added {quantity} x {item} to cart. Total: ₹{price}\n")

    except KeyError:
        print(" Error: Item not found in inventory.\n")

# BILL GENERATION
print("\n=== Final Bill ===")
total_price = sum(p[2] for p in cart)

try:
    if total_price == 0:
        raise ZeroDivisionError
    discount_percent = float(input("Enter discount percentage: "))
    discounted_total = total_price - (total_price * (discount_percent / 100))
    print(f"Total before discount: ₹{total_price}")
    print(f"Total after {discount_percent}% discount: ₹{discounted_total}")
except ZeroDivisionError:
    print(" Cannot apply discount because total price is zero.")

print("\nPurchased Items:")
for item_name, qty, price in cart:
    print(f"{item_name} - {qty} pcs - ₹{price}")
```

## Email Signup Validation


```python


class InvalidDomainError(Exception):
    pass

def validate_email(email):
    if "@" not in email:
        raise ValueError
    domain = email.split("@")[1]
    if domain.lower() != "example.com":
        raise InvalidDomainError
    return "Valid email"

# Main program
email = input("Enter your email: ").strip()

try:
    result = validate_email(email)
    print(result)
except ValueError:
    print("Email must contain '@' symbol.")
except InvalidDomainError:
    print("Email domain must be 'example.com'.")
except Exception as e:
    print(f"Unexpected error: {e}")
```
