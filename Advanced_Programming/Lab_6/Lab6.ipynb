# Q.1 University Course Registration System

```python
# Base class
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email

# Student class
class Student(Person):
    def __init__(self, name, email, student_id):
        super().__init__(name, email)
        self.student_id = student_id
        self.registered_courses = []

    def register_course(self, course):
        if course.add_student(self):
            self.registered_courses.append(course.course_code)
            print(f"{self.name} registered for {course.course_name}")
        else:
            print(f"{self.name} could not register for {course.course_name} (Full)")

# Faculty class
class Faculty(Person):
    def __init__(self, name, email, faculty_id):
        super().__init__(name, email)
        self.faculty_id = faculty_id
        self.assigned_courses = []

    def assign_course(self, course):
        self.assigned_courses.append(course.course_code)
        print(f"{self.name} assigned to course {course.course_name}")

# Course class
class Course:
    def __init__(self, course_code, course_name, capacity):
        self.course_code = course_code
        self.course_name = course_name
        self.capacity = capacity
        self.registered_students = []

    def add_student(self, student):
        if len(self.registered_students) < self.capacity:
            self.registered_students.append(student.student_id)
            University.total_students += 1
            return True
        return False

# University Controller class
class University:
    total_students = 0

    def __init__(self):
        self.all_courses = {}  # course_code: Course
        self.all_students = {}  # student_id: Student

    def add_course(self, course):
        self.all_courses[course.course_code] = course

    def add_student(self, student):
        self.all_students[student.student_id] = student

    def display_all_courses(self):
        print("\nAll Courses:")
        for course in self.all_courses.values():
            print(f"{course.course_code}: {course.course_name} - Registered: {len(course.registered_students)}/{course.capacity}")

    def display_all_students(self):
        print("\nAll Students:")
        for student in self.all_students.values():
            print(f"{student.student_id}: {student.name} - Courses: {student.registered_courses}")
```

```python
# Create University
uni = University()

# Create Courses
course1 = Course("CSE101", "Intro to CS", 2)
course2 = Course("MAT101", "Calculus", 3)
course3 = Course("PHY101", "Physics", 2)
course4 = Course("ENG101", "English", 2)

# Add Courses
for c in [course1, course2, course3, course4]:
    uni.add_course(c)

# Create Faculties
faculty1 = Faculty("Dr. Smith", "smith@uni.edu", "F001")
faculty2 = Faculty("Dr. Alice", "alice@uni.edu", "F002")
faculty3 = Faculty("Dr. John", "john@uni.edu", "F003")

# Assign Courses
faculty1.assign_course(course1)
faculty2.assign_course(course2)
faculty3.assign_course(course3)

# Create Students
students = [
    Student("Alice", "alice@student.com", "S001"),
    Student("Bob", "bob@student.com", "S002"),
    Student("Charlie", "charlie@student.com", "S003"),
    Student("David", "david@student.com", "S004"),
    Student("Eva", "eva@student.com", "S005")
]

# Add Students to University
for s in students:
    uni.add_student(s)

# Student Registrations
students[0].register_course(course1)
students[1].register_course(course1)
students[2].register_course(course1)  # Should fail (full)
students[2].register_course(course2)
students[3].register_course(course2)
students[4].register_course(course2)

# Display Results
uni.display_all_courses()
uni.display_all_students()
print(f"\nTotal Students Registered: {University.total_students}")
```

# Q.2 Hospital Patient Management System

```python
# Base class
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

# Doctor class
class Doctor(Person):
    def __init__(self, name, age, gender, doctor_id, specialization, available_slots):
        super().__init__(name, age, gender)
        self.doctor_id = doctor_id
        self.specialization = specialization
        self.available_slots = available_slots

    def book_slot(self, slot):
        if slot in self.available_slots:
            self.available_slots.remove(slot)
            return True
        else:
            raise ValueError(f"Slot {slot} is already booked or not available.")

# Patient class
class Patient(Person):
    def __init__(self, name, age, gender, patient_id, disease):
        super().__init__(name, age, gender)
        self.patient_id = patient_id
        self.disease = disease
        self.assigned_doctor = None
        self.appointment_slot = None

# Appointment class (Aggregation)
class Appointment:
    def __init__(self, appointment_id, doctor, patient, slot):
        self.appointment_id = appointment_id
        self.doctor = doctor
        self.patient = patient
        self.slot = slot

    def confirm_appointment(self):
        try:
            if self.doctor.book_slot(self.slot):
                self.patient.assigned_doctor = self.doctor.name
                self.patient.appointment_slot = self.slot
                return True
        except ValueError as e:
            print(f"Appointment Error: {e}")
        return False

# Hospital Controller
class Hospital:
    total_appointments = 0

    def __init__(self):
        self.all_doctors = {}  # doctor_id: Doctor
        self.all_patients = {}  # patient_id: Patient
        self.all_appointments = []  # list of Appointment objects

    def add_doctor(self, doctor):
        self.all_doctors[doctor.doctor_id] = doctor

    def add_patient(self, patient):
        self.all_patients[patient.patient_id] = patient

    def book_appointment(self, patient_id, doctor_id, slot):
        patient = self.all_patients.get(patient_id)
        doctor = self.all_doctors.get(doctor_id)
        if patient and doctor:
            appointment_id = f"APT{Hospital.total_appointments + 1}"
            appointment = Appointment(appointment_id, doctor, patient, slot)
            if appointment.confirm_appointment():
                self.all_appointments.append(appointment)
                Hospital.total_appointments += 1
                print(f"Appointment {appointment_id} confirmed for {patient.name} with Dr. {doctor.name} at {slot}")
            else:
                print(f"Failed to confirm appointment for {patient.name}")
        else:
            print("Invalid doctor or patient ID.")

    def display_all_appointments(self):
        print("\nAll Appointments:")
        for app in self.all_appointments:
            print(f"{app.appointment_id}: {app.patient.name} with Dr. {app.doctor.name} at {app.slot}")
```

```python
# Create Hospital
hosp = Hospital()

# Create Doctors
doc1 = Doctor("Dr. James", 45, "M", "D001", "Cardiology", ["10AM", "11AM"])
doc2 = Doctor("Dr. Emily", 38, "F", "D002", "Neurology", ["10AM", "1PM"])
doc3 = Doctor("Dr. Liam", 50, "M", "D003", "Orthopedics", ["9AM", "2PM"])

# Add Doctors
for doc in [doc1, doc2, doc3]:
    hosp.add_doctor(doc)

# Create Patients
patients = [
    Patient("Alice", 30, "F", "P001", "Headache"),
    Patient("Bob", 45, "M", "P002", "Back Pain"),
    Patient("Cathy", 29, "F", "P003", "Knee Injury"),
    Patient("David", 50, "M", "P004", "Heart Issues"),
    Patient("Eva", 33, "F", "P005", "Migraine"),
    Patient("Frank", 60, "M", "P006", "Fracture")
]

# Add Patients
for p in patients:
    hosp.add_patient(p)

# Book Appointments
hosp.book_appointment("P001", "D002", "10AM")
hosp.book_appointment("P002", "D003", "9AM")
hosp.book_appointment("P003", "D003", "2PM")
hosp.book_appointment("P004", "D001", "10AM")
hosp.book_appointment("P005", "D002", "10AM")  # Should fail (already booked)

# Display
hosp.display_all_appointments()
print(f"\nTotal Appointments: {Hospital.total_appointments}")
```
