### Q1. Denomination Calculator
Write a Python program to input an amount (integer) and determine the minimum number of notes required to make up that amount using denominations of 2000, 500, 200, 100, 50, 20, 10, 5, 2, and 1.  
#### Requirements
- Prompt user input
- Use conditionals and loops to calculate notes  
- Handle edge cases (e.g., 0 or negative inputs)


```python
def denomination_calculator():
    
    amount = int(input("Enter the amount: "))

    if (amount>0):
        for i in [2000, 500, 200, 50, 20, 5, 2, 1]:
            count = amount // i
            if (count>0):
                print(f"\t{count} notes of {i} denomination is required")
            amount %= i
    else: 
        print("Invalid input, can't be less than zero")

    
```

```python
denomination_calculator()
```

   ### Q.2 Calculator
  Build a menu-driven scientific calculator that performs operations like addition, subtraction, multiplication, division, and power for:  
     
   #### Requirements  
-	Float numbers  
-	Complex numbers  
  `Implement validation for invalid operations`


```python
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Error! Division by zero is not allowed."
    return x / y

def calculator():
    while True:
        print("\nSimple Python Calculator")
        print("1. Addition (+)")
        print("2. Subtraction (-)")
        print("3. Multiplication (*)")
        print("4. Division (/)")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == '5':
            print("Exiting the calculator. Goodbye!")
            break

        if choice in ['1', '2', '3', '4']:
            
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))

            if choice == '1':
                result = add(num1, num2)
                print(f"Result: {num1} + {num2} = {result}")

            elif choice == '2':
                result = subtract(num1, num2)
                print(f"Result: {num1} - {num2} = {result}")

            elif choice == '3':
                result = multiply(num1, num2)
                print(f"Result: {num1} * {num2} = {result}")

            elif choice == '4':
                result = divide(num1, num2)
                print(f"Result: {num1} / {num2} = {result}")

        else:
            print("Invalid input! Please enter a number between 1 and 5.")


```

```python
calculator()
```

### Q.3 Invoice Generator
Write a program that simulates an invoice generator for a small shop. It should:     
-	Accept item details: name, quantity, unit price  
-	Calculate item total and apply 18% GST (use Decimal for accuracy)  
-	Show the final bill with a breakdown  
`Use loops to accept multiple items. Use Decimal for calculations. Print formatted output using alignment`


```python
def invoice_generator():
    
    items= []
    final_bill=0
    
    while(1):
        
        # input the details
        name = str(input("Enter item name or -1 to stop: "))
        if (name=='-1'):
            break
            
        quantity = int(input("Enter the quantity: "))
        unit_price = int(input("Enter the unit price "))
        
        total_unit = quantity * unit_price
        total_taxed_unit = 0.18 * total_unit + total_unit
        
        # append
        items.append((name, quantity, unit_price, total_unit, total_taxed_unit))
        final_bill += total_taxed_unit
    
    print("Final Bill:\n")
    print(f"{'Name':<15}\t{'Quantity':<8}\t{'Unit Price':<10}\t{'Total Price':<12}\t{'Taxes per unit':<15}")

    for i in items:
        print(f"{i[0]:<15}\t{i[1]:<8}\t{i[2]:<10.2f}\t{i[3]:<12.2f}\t{i[4]:<15.2f}")

    print(f"\nNet amount: {final_bill:.2f}")


invoice_generator()
  
```

### Q. 4 Temperature Convertor
   Create a program that takes temperatures in Fahrenheit for 7 days, converts them to Celsius, and finds:  
-	Highest and lowest temperatures  
-	Average temperature  
-	Number of days below 20°C   
`Use the formula C = (F - 32) * 5/9.  Store results in a list. Use max(), min(), and average calculation. Loop to count specific values`


```python
def temp_finder():
    fah=[]
    cal=[]
    day=0
    print("Enter the temperature in fahrenheit:\n ")
    for i in range(0,7):
            temp = int(input(f"Day {i} temperature: "))
            fah.append(temp)
            cal.append((temp-32)*(5/9))
    
    # calculation
    count = 0
    sum_t = 0
    max_temp = max(cal)
    min_temp = min(cal)
    for i in range(7):
        sum_t += cal[i]
        if (cal[i]<20):
            count+=1
    avg_temp = sum_t/len(cal)
    
    print("\nTemperatures in Celsius:", [round(t,2) for t in cal])
    print("Highest temperature:", round(max_temp,2))
    print("Lowest temperature:", round(min_temp,2))
    print("Average temperature:", round(avg_temp, 2))
    print("Number of days below 20°C:", count)
```
