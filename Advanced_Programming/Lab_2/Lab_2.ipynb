# Q.1

Design and implement a Python program to manage hospital patient records using
appropriate data structures. Each patient has a unique patient_id (string), a name (string),
an age (integer), and a medical_history, which is a list of tuples. Each tuple in the medical
history represents a medical visit and should contain two elements: the date (in YYYYMM-DD format) and a diagnosis (string). Store the patient records in a dictionary, where
the key is the patient_id and the value is a tuple containing the patient's name, age, and
medical history list.
Your program should allow the following operations:
1. Add a new patient record to the system.
2. Update an existing patient’s medical history by appending a new (date, diagnosis)
tuple.
3. Display the details of all patients who are over the age of 60 and have at least one
diagnosis that includes the word "diabetes" (case-insensitive match).
4. Count and display how many patients have more than three medical diagnoses
in their history.

### Data types used:
string -> name, id  
int -> age  
[(tuple)] -> med_history = each tuple-> date, diagonises(str)

final-> dict{} -> key= id; val=(name, age, med_history)

## Solution

```python
final = {}  

while(1):
    stats = int(input("Enter 1 (Add)\nEnter 2 (Update)\nEnter 3 (Find 60+ w/ Diabetes)\nEnter 4 (Count >3 diagnoses)\nEnter 0 to stop: "))
    if (stats == 0):
        break

    if (stats == 1):    
        while(1):
            print("Enter 0 to exit or 1 to add details")
            status = int(input())
            if(status == 0):
                break
            patient_name = input("Enter patient name: ")
            patient_id = input("Enter patient id: ")
            age = int(input("Enter patient age: "))
            date = input("Enter date in YYYYMM-DD format: ")  # keep as string
            diagnosis = input("Enter type of diagnosis: ")

            elem = (date, diagnosis)
            medical_history = [elem]

            final[patient_id] = (patient_name, age, medical_history)
            print("Details added!\n")

    if (stats == 2):
        patient_id = input("Enter patient id: ")
        if patient_id in final.keys():
            date = input("Enter date in YYYYMM-DD format: ")
            diagnosis = input("Enter type of diagnosis: ")
            elem = (date, diagnosis)

            patient_name, age, medical_history = final[patient_id]
            medical_history.append(elem)
            final[patient_id] = (patient_name, age, medical_history)

            print("Details updated!\n")
        else:
            print("Patient ID not found!\n")

    if (stats == 3):
        for i in final.keys():
            name, age, history = final[i]
            if age > 60 and any("diabetes" in diag.lower() for (_, diag) in history):
                print(f"\nFull Record: {final[i]}")
                print(f"Name: {name}")
                print(f"Age: {age}")
                print(f"Medical History: {history}\n")

    if (stats == 4):
        count = 0
        for i in final.keys():
            if len(final[i][2]) > 3:
                count += 1
        print(f"{count} patient(s) have more than 3 diagnoses.\n")
```

# Q 2
Develop a Python program to assist a university in planning and analyzing course
schedules and student enrollments. Each course is uniquely identified by a course_code
(e.g., "CS101") and includes the following details: the instructor (string), a schedule
represented as a list of tuples, where each tuple contains the day and time of a session
(e.g., ("Mon", "10:00AM")), and a list of enrolled_students (names as strings). All course
data should be stored in a dictionary, with the course_code as the key and the value
being a tuple containing the instructor's name, schedule list, and the list of enrolled
students.
Your program should provide functionality to:
1. Add new courses or update the schedule of existing courses.
2. Detect and report scheduling conflicts for instructors—cases where an
instructor is assigned to more than one course at the same time on the same day.
3. Generate a new dictionary that maps each student name to a list of course
codes they are enrolled in.
4. Identify and display the names of all students who are enrolled in more than
three courses.

```python
courses = {}

while True:
    print("1. Add/Update Course")
    print("2. Check Instructor Scheduling Conflicts")
    print("3. Generate Student-Course Map")
    print("4. Find Students Enrolled in > 3 Courses")
    print("0. Exit")

    choice = input("Enter your choice: ")

    if choice == '0':
        print("Exiting program. Goodbye!")
        break

    elif choice == '1':
        course_code = input("Enter course code (e.g., CS101): ")
        instructor = input("Enter instructor name: ")
        
        schedule = []
        print("Enter schedule entries (day and time). Type 'done' to finish:")
        while True:
            day = input("Day (Mon/Tue/... or 'done' to finish): ")
            if day.lower() == 'done':
                break
            time = input("Time (e.g., 10:00AM): ")
            schedule.append((day, time))
        
        enrolled_students = []
        print("Enter student names. Type 'done' to finish:")
        while True:
            student = input("Student name (or 'done' to finish): ")
            if student.lower() == 'done':
                break
            enrolled_students.append(student)

        if course_code in courses:
            print(f"Updating existing course {course_code}.")
        else:
            print(f"Adding new course {course_code}.")

        courses[course_code] = (instructor, schedule, enrolled_students)
        print("Course added/updated successfully.\n")

    elif choice == '2':
        instructor_schedule = {}
        conflicts_found = False
        for course_code, value in courses.items():
            instructor, schedule, students = value
            for session in schedule:
                key = (instructor, session)
                if key in instructor_schedule:
                    print(f" Conflict: {instructor} has a clash at {session} between courses {instructor_schedule[key]} and {course_code}")
                    conflicts_found = True
                else:
                    instructor_schedule[key] = course_code
        if not conflicts_found:
            print("No instructor scheduling conflicts found.\n")

    elif choice == '3':
        student_map = {}
        for course_code, value in courses.items():
            instructor, schedule, students = value
            for student in students:
                if student not in student_map:
                    student_map[student] = []
                student_map[student].append(course_code)
        
        print("\nStudent-Course Map:")
        for student, course_list in student_map.items():
            print(f"{student}: {course_list}")

    elif choice == '4':
        student_map = {}
        for course_code, value in courses.items():
            instructor, schedule, students = value
            for student in students:
                if student not in student_map:
                    student_map[student] = []
                student_map[student].append(course_code)

        found = False
        print("\nStudents enrolled in more than 3 courses:")
        for student, course_list in student_map.items():
            if len(course_list) > 3:
                print(f"{student}: {course_list}")
                found = True
        if not found:
            print("No students enrolled in more than 3 courses.")

    else:
        print("Invalid choice. Please try again.")
```
