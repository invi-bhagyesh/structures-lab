```python
def text_data_compression():
    try:
        paragraph = input("Enter a paragraph: ")
        if not isinstance(paragraph, str):
            raise AttributeError("Input must be a string.")

        # Normalize and split paragraph into words (lowercase, strip punctuation)
        import string
        translator = str.maketrans('', '', string.punctuation)
        words = paragraph.translate(translator).lower().split()

        # a. Count frequency of each word
        freq_dict = {}
        for word in words:
            freq_dict[word] = freq_dict.get(word, 0) + 1

        # b. Convert dictionary into list of tuples (word, frequency)
        freq_list = list(freq_dict.items())

        # c. Sort list by frequency descending
        freq_list.sort(key=lambda x: x[1], reverse=True)

        print("\nWord frequencies sorted by descending order:")
        for idx, (word, freq) in enumerate(freq_list):
            print(f"{idx}: {word} -> {freq}")

        # d. Allow user to retrieve frequency of a particular word or by index
        while True:
            choice = input("\nRetrieve frequency by 'word', 'index', or type 'exit' to quit: ").strip().lower()
            if choice == 'exit':
                break
            elif choice == 'word':
                try:
                    search_word = input("Enter the word to find frequency: ").lower()
                    freq = freq_dict[search_word]
                    print(f"The word '{search_word}' appears {freq} times.")
                except KeyError:
                    print(f"Word '{search_word}' not found in the paragraph.")
            elif choice == 'index':
                try:
                    index = int(input("Enter the index to retrieve (0-based): "))
                    word, freq = freq_list[index]
                    print(f"Index {index}: '{word}' appears {freq} times.")
                except IndexError:
                    print("Invalid index! Please enter a valid index.")
                except ValueError:
                    print("Please enter a numeric index.")
            else:
                print("Invalid option. Please enter 'word', 'index', or 'exit'.")

    except AttributeError as e:
        print("Error:", e)

```

```python
text_data_compression()
```

```python
class NegativeSalaryError(Exception):
    """Custom exception raised when final salary is negative."""
    pass

def employee_payroll_system():
    # a. Employee dictionary: ID -> (name, base_salary, department)
    employees = {
        'E001': ('Alice Johnson', 5000, 'HR'),
        'E002': ('Bob Smith', 6000, 'IT'),
        'E003': ('Charlie Lee', 5500, 'Finance'),
    }

    while True:
        emp_id = input("\nEnter Employee ID to search (or type 'exit' to quit): ").strip()
        if emp_id.lower() == 'exit':
            print("Exiting payroll system.")
            break

        try:
            # b. Search employee by ID
            if emp_id not in employees:
                raise KeyError(f"Employee ID '{emp_id}' does not exist.")
            name, base_salary, department = employees[emp_id]
            print(f"Employee found: {name}, Department: {department}, Base Salary: ${base_salary}")

            # c. Get salary adjustment from user
            adj_input = input("Enter salary adjustment (bonus positive / deduction negative): ").strip()
            try:
                adjustment = float(adj_input)
            except ValueError:
                raise ValueError("Salary adjustment must be a numeric value.")

            final_salary = base_salary + adjustment
            if final_salary < 0:
                raise NegativeSalaryError(f"Final salary cannot be negative. Calculated: {final_salary}")

            # d. Print salary slip
            print("\n--- Salary Slip ---")
            print(f"Employee ID   : {emp_id}")
            print(f"Name          : {name}")
            print(f"Department    : {department}")
            print(f"Base Salary   : ${base_salary:.2f}")
            print(f"Adjustment    : ${adjustment:.2f}")
            print(f"Final Salary  : ${final_salary:.2f}")
            print("------------------")

        except KeyError as ke:
            print("Error:", ke)
        except ValueError as ve:
            print("Error:", ve)
        except NegativeSalaryError as ne:
            print("Error:", ne)

```

```python
employee_payroll_system()
```
